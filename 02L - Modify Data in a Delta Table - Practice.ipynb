### Task 1: Insert a new employee

Add a new row to `new_employees` with the following values:

- **FirstName:** Priya
- **Country:** India
- **Role:** Data Engineer

### Task 2: Update a role

Aiden has been promoted. Update his role from `Data Analyst` to `Senior Data Analyst`.

```python
%sql
UPDATE new_employees
SET Role = 'Senior Data Analyst'
WHERE FirstName = 'Aiden';
```

<details>
<summary>Hint</summary>

<pre><code>UPDATE new_employees
SET Role = 'Senior Data Analyst'
WHERE FirstName = 'Aiden';</code></pre>

</details>

Verify the update. Aiden's role should now show `Senior Data Analyst`.

```python
%sql
SELECT * FROM new_employees WHERE FirstName = 'Aiden';
```

### Task 4: Check the version history

Use `DESCRIBE HISTORY` to see all the changes you just made to `new_employees`.
